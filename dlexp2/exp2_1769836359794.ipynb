# AND/OR gate

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y_and = np.array([0,0,0,1])
y_or  = np.array([0,1,1,1])

def step(z):
    return np.where(z >= 0, 1, 0)

def train_perceptron(X, y, lr=0.1, epochs=20):
    w = np.zeros(X.shape[1])
    b = 0
    for _ in range(epochs):
        for xi, yi in zip(X, y):
            z = np.dot(w, xi) + b
            y_hat = step(z)
            error = yi - y_hat
            w += lr * error * xi
            b += lr * error
    return w, b

w_and, b_and = train_perceptron(X, y_and)
w_or,  b_or  = train_perceptron(X, y_or)

print("AND weights:", w_and, "bias:", b_and)
print("OR weights:", w_or, "bias:", b_or)

import numpy as np
import matplotlib.pyplot as plt

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y_and = np.array([0,0,0,1])
y_or  = np.array([0,1,1,1])

def plot_safe_boundary(X, y, boundary_value, title)
    plt.figure(figsize=(5,5))

   
    for i in range(len(X)):
        if y[i] == 1:
            plt.scatter(X[i,0], X[i,1], marker='o', s=150, label='Class 1' if i==1 else "")
        else:
            plt.scatter(X[i,0], X[i,1], marker='x', s=150, label='Class 0' if i==0 else "")

    
    x_vals = np.linspace(-0.2, 1.2, 100)
    y_vals = boundary_value - x_vals
    plt.plot(x_vals, y_vals, 'r--', linewidth=2, label="Separating Boundary")

    plt.xticks([0,1])
    plt.yticks([0,1])
    plt.xlabel("Input X1")
    plt.ylabel("Input X2")
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

plot_safe_boundary(X, y_and, boundary_value=1.5,
                   title="AND Gate ")

plot_safe_boundary(X, y_or, boundary_value=0.5,
                   title="OR Gate ")
```

# Binary Cross Entropy


```python
import pandas as pd
import torch
import matplotlib.pyplot as plt

data = pd.read_csv("Pumpkin_Seeds_Dataset.csv")

print("Columns:", data.columns)

data["Class"] = data["Class"].map({
    "Çerçevelik": 1,
    "Ürgüp": 0
})

X = torch.tensor(
    data.drop("Class", axis=1).values,
    dtype=torch.float32
)

y = torch.tensor(
    data["Class"].values,
    dtype=torch.float32
).view(-1, 1)

print("Dataset shape:", X.shape)

W1 = torch.randn(X.shape[1], 8)
W2 = torch.randn(8, 1)

# Forward propagation
z1 = X @ W1
a1 = torch.relu(z1)
z2 = a1 @ W2
y_hat = torch.sigmoid(z2)

# Binary Cross Entropy Loss 
epsilon = 1e-8  # avoid log(0)
bce_loss = -(y * torch.log(y_hat + epsilon) +
             (1 - y) * torch.log(1 - y_hat + epsilon))

loss = bce_loss.mean()

print("Average BCE Loss on Pumpkin Seeds Dataset:", loss.item())

y_true = y.numpy().flatten()
y_pred = y_hat.detach().numpy().flatten()

plt.figure(figsize=(7,5))


plt.scatter(range(len(y_pred)), y_pred, c=y_true, cmap='bwr', alpha=0.6)
plt.axhline(0.5, linestyle='--', color='black', label='Decision Boundary')

plt.xlabel("Sample Index")
plt.ylabel("Predicted Probability (Class = 1)")
plt.title("Pumpkin Seeds Dataset\nPredicted Probabilities (Forward Pass)")
plt.colorbar(label="True Class (0 = Ürgüp, 1 = Çerçevelik)")
plt.legend()
plt.grid(True)
plt.show()

loss_per_sample = bce_loss.detach().numpy().flatten()

plt.figure(figsize=(7,5))
plt.plot(loss_per_sample, color='purple', linewidth=1)
plt.xlabel("Sample Index")
plt.ylabel("Binary Cross Entropy Loss")
plt.title("Binary Cross Entropy Loss per Sample\n(Pumpkin Seeds Dataset)")
plt.grid(True)
plt.show()

```

# Categorical Cross Entropy


```python
import torch
import numpy as np
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt

iris = load_iris()
X = iris.data
y = iris.target  

X = StandardScaler().fit_transform(X)

X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)

print("Dataset shape:", X.shape)
input_dim = 4
hidden_dim = 5
output_dim = 3

W1 = torch.randn(input_dim, hidden_dim)
W2 = torch.randn(hidden_dim, output_dim)

# Forward propagation
z1 = X @ W1
a1 = torch.relu(z1)
logits = a1 @ W2
# Softmax implementation 
exp_scores = torch.exp(logits - torch.max(logits, dim=1, keepdim=True)[0])
softmax = exp_scores / torch.sum(exp_scores, dim=1, keepdim=True)
# Categorical Cross Entropy 
N = X.shape[0]
log_probs = -torch.log(softmax[torch.arange(N), y])
loss = log_probs.mean()

print("Softmax output (first 5 samples):\n", softmax[:5])
print("Categorical Cross Entropy Loss:", loss.item())

softmax_np = softmax.detach().numpy()

plt.figure(figsize=(8,5))
plt.plot(softmax_np[:30, 0], label="Class 0 (Setosa)")
plt.plot(softmax_np[:30, 1], label="Class 1 (Versicolor)")
plt.plot(softmax_np[:30, 2], label="Class 2 (Virginica)")
plt.xlabel("Sample Index")
plt.ylabel("Predicted Probability")
plt.title("Softmax Output Probabilities (First 30 Iris Samples)")
plt.legend()
plt.grid(True)
plt.show()

y_pred = torch.argmax(softmax, dim=1)

# Convert to numpy
y_true_np = y.detach().numpy()
y_pred_np = y_pred.detach().numpy()

cm = confusion_matrix(y_true_np, y_pred_np)

print("Confusion Matrix:\n", cm)

plt.figure(figsize=(6,5))
plt.imshow(cm)
plt.colorbar()

plt.xticks(range(3), iris.target_names)
plt.yticks(range(3), iris.target_names)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix (Iris Dataset)")

# Annotate values
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j],
                 ha="center", va="center", fontsize=12)

plt.show()



```

# MAE/MSE

```python
import torch
import pandas as pd
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

data = pd.read_csv("housing.csv")

data = data.drop("ocean_proximity", axis=1)

data = data.dropna()

y = data["median_house_value"].values
X = data.drop("median_house_value", axis=1).values

X = StandardScaler().fit_transform(X)

X = torch.tensor(X, dtype=torch.float32)
y_true = torch.tensor(y, dtype=torch.float32)

print("Dataset shape:", X.shape)


W = torch.randn(X.shape[1])
b = torch.randn(1)

y_pred = X @ W + b

mse = torch.mean((y_true - y_pred) ** 2)
mae = torch.mean(torch.abs(y_true - y_pred))

print("MSE:", mse.item())
print("MAE:", mae.item())

y_true_np = y_true.detach().numpy()
y_pred_np = y_pred.detach().numpy()

plt.figure(figsize=(6,6))
plt.scatter(y_true_np, y_pred_np, alpha=0.4)
plt.xlabel("Actual House Price")
plt.ylabel("Predicted House Price")
plt.title("Actual vs Predicted House Prices")
plt.grid(True)

min_val = min(y_true_np.min(), y_pred_np.min())
max_val = max(y_true_np.max(), y_pred_np.max())
plt.plot([min_val, max_val], [min_val, max_val], 'r--')
plt.show()

abs_error = torch.abs(y_true - y_pred).detach().numpy()

plt.figure(figsize=(7,5))
plt.hist(abs_error, bins=40)
plt.xlabel("Absolute Error")
plt.ylabel("Frequency")
plt.title("Distribution of Absolute Errors (MAE)")
plt.grid(True)
plt.show()

residuals = y_true_np - y_pred_np

plt.figure(figsize=(7,5))
plt.scatter(y_pred_np, residuals, alpha=0.4)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Predicted House Price")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Plot (Housing Regression)")
plt.grid(True)
plt.show()

```

# Stochastic, Mini batch and Full batch gradient descent

```python
import torch
import time
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler

X, y = load_breast_cancer(return_X_y=True)

X = StandardScaler().fit_transform(X)
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float32).view(-1, 1)

W = torch.randn(X.shape[1], 1)
def bce_loss(y_hat, y):
    eps = 1e-8
    return -(y * torch.log(y_hat + eps) +
             (1 - y) * torch.log(1 - y_hat + eps)).mean()

# Stochastic Gradient Descent
start = time.time()
for i in range(len(X)):
    xi = X[i:i+1]
    yi = y[i:i+1]
    y_hat = torch.sigmoid(xi @ W)
    loss = bce_loss(y_hat, yi)
sgd_time = time.time() - start

# Mini-Batch Gradient Descent
batch_size = 32
start = time.time()
for i in range(0, len(X), batch_size):
    xb = X[i:i+batch_size]
    yb = y[i:i+batch_size]
    y_hat = torch.sigmoid(xb @ W)
    loss = bce_loss(y_hat, yb)
mini_batch_time = time.time() - start

# Full Batch Gradient Descent
start = time.time()
y_hat = torch.sigmoid(X @ W)
loss = bce_loss(y_hat, y)
full_batch_time = time.time() - start
methods = ["Stochastic", "Mini-Batch", "Full Batch"]
times = [sgd_time, mini_batch_time, full_batch_time]

plt.figure(figsize=(7,5))
plt.bar(methods, times)
plt.ylabel("Time (seconds)")
plt.title("Training Time Comparison (Breast Cancer Dataset)")
plt.grid(axis='y')
plt.show()
```
